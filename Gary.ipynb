read and display file genshin_impact.pdf as a dataframe

```python
import pdfplumber
import pandas as pd
import re

pdf_path = '/content/genshin_impact.pdf'
rows = []
header_line = ""

with pdfplumber.open(pdf_path) as pdf:
    for i, page in enumerate(pdf.pages):
        text = page.extract_text()
        if text:
            lines = text.strip().split('\n')
            for line in lines:
                # If it's a page label line, skip it
                if line.startswith("--- Page") or not line.strip():
                    continue

                # Grab the header line dynamically on first page
                if i == 0 and not header_line:
                    header_line = line
                    continue

                # Simple parser based on known structure:
                # format: [index][name] [rarity][weapon] [element] [model] [region] ... [pulled_count]
                # Let's tokenize using regex to capture bracketed lists, dates, and separate tokens cleanly.
                tokens = re.findall(r'\d+\S+|\d+\s+[^\d\s]+|[^\s\[\]]+|\[[^\]]*\]', line)
                if tokens:
                    rows.append(tokens)

# Create a simplified DataFrame
# We'll use the extracted raw columns and pad/truncate to align rows neatly
max_cols = max(len(r) for r in rows) if rows else 0
columns = [f"Col_{j}" for j in range(max_cols)]

# Map the most important known headers
known_headers = ["ID_Name", "Rarity_Weapon", "Element", "Model_Region", "Release_Date", "Days_Since", "Banners", "Roles", "Stats_Constellations", "Pulled_Count"]
for k in range(len(known_headers)):
    if k < len(columns):
        columns[k] = known_headers[k]

df = pd.DataFrame([r + [''] * (max_cols - len(r)) for r in rows], columns=columns)
display(df.head(15))




```

compare in a bar chart rarity 5 to rarity 4 characters ATK stats

```python
import re
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Let's extract character names along with their Rarity and Base ATK
character_data = []

for idx, row in df.iterrows():
    raw_name = str(row['ID_Name'])
    # Clean the index prefix from name (e.g. '0lynette' -> 'lynette', '102albedo' -> 'albedo')
    name_clean = re.sub(r'^\d+', '', raw_name).strip().capitalize()

    raw_rarity_weapon = str(row['Rarity_Weapon'])
    rarity_match = re.search(r'\d', raw_rarity_weapon)
    if not rarity_match:
        continue
    rarity = int(rarity_match.group())

    # Extract all pure numbers in the row to locate the base ATK
    numbers = []
    for col in df.columns:
        val = str(row[col])
        if val.isdigit():
            numbers.append(int(val))

    # Match our known ATK bounds (100 - 450)
    atks = [n for n in numbers if 100 <= n <= 450]

    if atks and name_clean:
        character_data.append({
            'Name': name_clean,
            'Rarity': f'{rarity}-Star',
            'Base ATK': atks[0]
        })

char_df = pd.DataFrame(character_data).drop_duplicates(subset=['Name'])

# Get top 5 for 5-Star
top_5_star = char_df[char_df['Rarity'] == '5-Star'].nlargest(5, 'Base ATK')

# Get top 5 for 4-Star
top_4_star = char_df[char_df['Rarity'] == '4-Star'].nlargest(5, 'Base ATK')

# Combine them for plotting
top_chars_df = pd.concat([top_5_star, top_4_star])

# Plot the comparison
plt.figure(figsize=(12, 6))
sns.barplot(
    data=top_chars_df,
    x='Name',
    y='Base ATK',
    hue='Rarity',
    palette='Set2'
)

plt.title('Top 5 Rarity 5 vs Top 5 Rarity 4 Characters by Base ATK', fontsize=14, weight='bold')
plt.xlabel('Character Name')
plt.ylabel('Base ATK')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

the top 5 for each being compared by ATK stats still has 5-stars be higher than 4-stars

combine both top 5's into their own respecive average, 5-star vs 4-star

```python
# Compute the average of the top 5 characters for each rarity group
top_averages = top_chars_df.groupby('Rarity')['Base ATK'].mean().reset_index()

# Plot the comparative averages
plt.figure(figsize=(8, 6))
sns.barplot(
    data=top_averages,
    x='Rarity',
    y='Base ATK',
    hue='Rarity',
    palette='Set2',
    legend=False
)

# Add exact value labels on top of the bars
for index, row in top_averages.iterrows():
    plt.text(
        index,
        row['Base ATK'] + 5,
        f"{row['Base ATK']:.2f}",
        color='black',
        ha="center",
        weight='bold'
    )

plt.title('Average Base ATK: Top 5 Rarity 5 vs Top 5 Rarity 4 Characters', fontsize=12, weight='bold')
plt.xlabel('Rarity Group')
plt.ylabel('Average Base ATK')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0, max(top_averages['Base ATK']) * 1.15)
plt.tight_layout()
plt.show()
```

Re-create the bar chart with the SEM(standard error mean) for a 95% confidience ( 5% significance) using only the atk stats for the average stats of the top 5-stars vs top 5 4-stars


```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Group the top 5 characters data by Rarity and compute mean and SEM for Base ATK
top_stats = top_chars_df.groupby('Rarity')['Base ATK'].agg(['mean', 'sem']).reset_index()

# For sample sizes as small as 5, we use 1.96 * SEM for the 95% Confidence Interval error margin
top_stats['ci_error'] = 1.96 * top_stats['sem']

# Plotting
plt.figure(figsize=(8, 6))
plt.bar(
    top_stats['Rarity'],
    top_stats['mean'],
    yerr=top_stats['ci_error'],
    capsize=10,
    color=sns.color_palette('Set2', len(top_stats)),
    edgecolor='gray',
    alpha=0.9,
    width=0.4
)

# Annotate the averages and confidence intervals inside/on the bars
for idx, row in top_stats.iterrows():
    plt.text(
        idx,
        row['mean'] / 2,
        f"{row['mean']:.2f}\n(±{row['ci_error']:.2f})",
        color='black',
        ha="center",
        va="center",
        weight='bold',
        fontsize=11
    )

plt.title('Average Base ATK of Top 5 Characters (with 95% CI)', fontsize=13, weight='bold')
plt.xlabel('Rarity Group', fontsize=12)
plt.ylabel('Average Base ATK', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0, max(top_stats['mean'] + top_stats['ci_error']) * 1.25)
plt.tight_layout()
plt.show()
```

its statistacly significant enough to choose 5-stars

```python
# Clean and extract Base DEF along with character names
def_data = []

for idx, row in df.iterrows():
    raw_name = str(row['ID_Name'])
    name_clean = re.sub(r'^\d+', '', raw_name).strip().capitalize()

    raw_rarity_weapon = str(row['Rarity_Weapon'])
    rarity_match = re.search(r'\d', raw_rarity_weapon)
    if not rarity_match:
        continue
    rarity = int(rarity_match.group())

    numbers = []
    for col in df.columns:
        val = str(row[col])
        if val.isdigit():
            numbers.append(int(val))

    # Filter sensible DEF stats (typically 500-1000)
    defs = [n for n in numbers if 500 <= n <= 1000]

    if defs and name_clean:
        def_data.append({
            'Name': name_clean,
            'Rarity': f'{rarity}-Star',
            'Base DEF': defs[0]
        })

def_df = pd.DataFrame(def_data).drop_duplicates(subset=['Name'])

# Get top 5 for 5-Star
top_5_star_def = def_df[def_df['Rarity'] == '5-Star'].nlargest(5, 'Base DEF')

# Get top 5 for 4-Star
top_4_star_def = def_df[def_df['Rarity'] == '4-Star'].nlargest(5, 'Base DEF')

# Combine them to compute the average
top_def_combined = pd.concat([top_5_star_def, top_4_star_def])
top_def_averages = top_def_combined.groupby('Rarity')['Base DEF'].mean().reset_index()

# Plot the comparison bar chart
plt.figure(figsize=(8, 6))
sns.barplot(
    data=top_def_averages,
    x='Rarity',
    y='Base DEF',
    hue='Rarity',
    palette='Set2',
    legend=False
)

# Add exact average values on top of the bars
for index, row in top_def_averages.iterrows():
    plt.text(
        index,
        row['Base DEF'] + 15,
        f"{row['Base DEF']:.2f}",
        color='black',
        ha="center",
        weight='bold'
    )

plt.title('Average Base DEF: Top 5 Rarity 5 vs Top 5 Rarity 4 Characters', fontsize=12, weight='bold')
plt.xlabel('Rarity Group')
plt.ylabel('Average Base DEF')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0, max(top_def_averages['Base DEF']) * 1.15)
plt.tight_layout()
plt.show()
```

Re-create the bar chart with the SEM(standard error mean) for a 95% confidience ( 5% significance) using only the atk stats for the average stats of the top 5-stars vs top 5 4-stars

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# Group the top 5 characters by Rarity and compute mean and SEM for Base DEF
top_def_stats = top_def_combined.groupby('Rarity')['Base DEF'].agg(['mean', 'sem']).reset_index()

# Calculate the error representing a 95% Confidence Interval (1.96 * SEM)
top_def_stats['ci_error'] = 1.96 * top_def_stats['sem']

# Plotting
plt.figure(figsize=(8, 6))
plt.bar(
    top_def_stats['Rarity'],
    top_def_stats['mean'],
    yerr=top_def_stats['ci_error'],
    capsize=10,
    color=sns.color_palette('Set2', len(top_def_stats)),
    edgecolor='gray',
    alpha=0.9,
    width=0.4
)

# Annotate the averages and confidence intervals on the bars
for idx, row in top_def_stats.iterrows():
    plt.text(
        idx,
        row['mean'] / 2,
        f"{row['mean']:.2f}\n(±{row['ci_error']:.2f})",
        color='black',
        ha="center",
        va="center",
        weight='bold',
        fontsize=11
    )

plt.title('Average Base DEF of Top 5 Characters (with 95% CI)', fontsize=13, weight='bold')
plt.xlabel('Rarity Group', fontsize=12)
plt.ylabel('Average Base DEF', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.ylim(0, max(top_def_stats['mean'] + top_def_stats['ci_error']) * 1.25)
plt.tight_layout()
plt.show()
```

apply the statistical test chisquare from spicy.stats, comparing the ATK and the DEF

```python
from scipy.stats import chi2_contingency
import numpy as np

# Construct a 2x2 contingency table comparing ATK and DEF across rarities
# Rows: [4-Star, 5-Star]
# Columns: [Average ATK, Average DEF]
contingency_table = np.array([
    [top_stats.loc[top_stats['Rarity'] == '4-Star', 'mean'].values[0], top_def_stats.loc[top_def_stats['Rarity'] == '4-Star', 'mean'].values[0]],
    [top_stats.loc[top_stats['Rarity'] == '5-Star', 'mean'].values[0], top_def_stats.loc[top_def_stats['Rarity'] == '5-Star', 'mean'].values[0]]
])

# Run the Chi-Square Test of Independence
chi2_stat, p_val, dof, expected = chi2_contingency(contingency_table)

print("Contingency Table (Rarity vs Stats):")
print(f"4-Star: ATK = {contingency_table[0,0]:.2f}, DEF = {contingency_table[0,1]:.2f}")
print(f"5-Star: ATK = {contingency_table[1,0]:.2f}, DEF = {contingency_table[1,1]:.2f}\n")

print(f"Chi-Square Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_val:.4f}")
print(f"Degrees of Freedom: {dof}")

if p_val < 0.05:
    print("The association between Rarity and Stat types (ATK vs DEF) is statistically significant (p < 0.05).")
else:
    print("The association between Rarity and Stat types (ATK vs DEF) is not statistically significant (p >= 0.05).")
```

apply the statistical test chisquare from spicy.stats, comparing the ATK and the DEF and give seperate percentages for the ATK and DEF

```python
import pandas as pd

# Calculate the total of (ATK + DEF) for top 5 characters in each rarity
top_4star_atk = top_stats.loc[top_stats['Rarity'] == '4-Star', 'mean'].values[0]
top_4star_def = top_def_stats.loc[top_def_stats['Rarity'] == '4-Star', 'mean'].values[0]
top_4star_total = top_4star_atk + top_4star_def

top_5star_atk = top_stats.loc[top_stats['Rarity'] == '5-Star', 'mean'].values[0]
top_5star_def = top_def_stats.loc[top_def_stats['Rarity'] == '5-Star', 'mean'].values[0]
top_5star_total = top_5star_atk + top_5star_def

# Compute percentages
pct_4star_atk = (top_4star_atk / top_4star_total) * 100
pct_4star_def = (top_4star_def / top_4star_total) * 100

pct_5star_atk = (top_5star_atk / top_5star_total) * 100
pct_5star_def = (top_5star_def / top_5star_total) * 100

print("=== Stat Distribution Percentages (Top 5 Characters) ===")
print(f"4-Star: Base ATK represents {pct_4star_atk:.2f}% and Base DEF represents {pct_4star_def:.2f}% of the overall stat total ({top_4star_total:.2f}).")
print(f"5-Star: Base ATK represents {pct_5star_atk:.2f}% and Base DEF represents {pct_5star_def:.2f}% of the overall stat total ({top_5star_total:.2f}).")
```
